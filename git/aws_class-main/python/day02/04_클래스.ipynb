```python
# 2차원 좌표평면의 점을 나타내는 Point 클래스를 선언


# Point 클래스의 객체를 생성
class Point:
    # 클래스 변수
    name : str = "포인트 클래스"

    # 생성자
    def __init__(self, x:int = 0, y:int = 0):
        # 인스턴스 변수
        self.__x = x
        self.__y = y

    # 함수
    def print(self):
        print(f"({self.__x}, {self.__y})")

    # __x의 getter
    @property 
    def x(self):
        return self.__x

     # __x의 setter
    @x.setter
    def x(self, x):
        self.__x = x

    @property
    def y(self):
        return self.__y

    @y.setter
    def y(self, y):
        self.__y = y

print(Point.name)
p1 = Point(0,0)
p2 = Point(1,1)
p3 = Point()
p1.print()
p2.print()
p3.x = 100
p3.y = 100
p3.print()
```

```python
# 은행 계좌 클래스
# 소유자(owner), 잔액(balance)
# 기능 : 출금(withdraw), 입금(deposit)

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        print(f"{self.owner}님의 계좌가 개설되었습니다.")

    def deposit(self, amount):
        self.balance += amount
        print(f"{self.owner}님의 계좌에 {amount}원이 입금되었습니다.")

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"{self.owner}님의 계좌에서 {amount}원이 출금되었습니다.")
        else:
            print(f"{self.owner}님의 계좌에 잔액이 부족합니다.")


# 홍길동이 계좌 개설
hong_account = BankAccount("홍길동")
lim_account = BankAccount("임꺽정")
go_account = BankAccount("고길동")
# 100000(십만) 입금
hong_account.deposit(100000)
lim_account.deposit(200000)
go_account.deposit(10000000)
# 50000(5만) 출금
hong_account.withdraw(50000)
lim_account.withdraw(1000000)
go_account.withdraw(5000000)
# 70000(7만) 출금 => 잔액 부족 알림
hong_account.withdraw(70000)
```

```python
# 도서 대여 시스템에서 사용할 도서 클래스 선언 및 활용
# 도서 클래스
# 저자, 도서명, 대출 가능 여부(is_borrow)
# 기능 : 대출, 반납, 정보확인
class Book:
    # 생성자
    def __init__(self, title : str, author : str):
        self.title = title
        self.author = author
        self.is_borrow = True
       
        
    # 대출
    def borrow_book(self, user_name : str):
        if self.is_borrow:
            self.is_borrow = False
            self.user_name = user_name
            print(f"{self.user_name}님이 '{self.title}'을 대출했습니다.")
        else:
            print(f"'{self.title}'은 이미 대출 중인 도서라 대출이 불가능합니다.")
    # 반납
    def return_book(self):
        if not self.is_borrow:
            print(f"{self.user_name}님이 '{self.title}'을 반납했습니다. 해당 도서는 대출이 가능합니다.")
            self.is_borrow = True
            self.user_name = ""
        else:
            print(f"'{self.title}'은 현재 대출 중인 상태가 아닙니다.")
    # 정보확인
    def show_info(self):
        status = "대출 가능" if self.is_borrow else "대출 중"
        print("-" * 20)
        print(f"제목 : {self.title}")
        print(f"저자 : {self.author}")
        print(f"대출여부 : {status}")
        print("-" * 20)

book1 = Book("파이썬 정석", "홍길동")
book2 = Book("AI 원리", "고길동")

book1.borrow_book("회원1") # 회원1님이 파이썬 정석을 대출했습니다.
book1.borrow_book("회원2") # 이미 대출 중인 도서라 대출이 불가능합니다.
book1.return_book() # 회원1님이 파이썬 정석을 반납했습니다. 해당 도서는 대출이 가능합니다.

book2.borrow_book("회원2")

book1.borrow_book("회원1")
book1.borrow_book("회원3")

book1.show_info()
book2.show_info()
# 제목 : 파이썬 정석
# 저자 : 홍길동
# 대출여부 : 대출 가능 # 이미 대출 중인 경우 대출 중
```
