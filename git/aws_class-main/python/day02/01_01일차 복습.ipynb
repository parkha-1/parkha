```python
# 정수를 입력받아 입력받은 정수가 홀수인지 짝수인지 출력하는 코드
num = input("정수를 입력하세요 : ")
num = int(num) # 문자열을 정수로 변환

if num % 2 == 0:
    print(f"{num}는 짝수")
else:
    print(f"{num}는 홀수")
print(f"{num}는 {"짝수" if num % 2 == 0 else "홀수"}")
```

```python
# 1부터 10까지의 합을 출력하는 예제
min, max, sum = 1, 10, 0
for i in range(min, max + 1):
    sum = sum + i # sum += {
print(f"{min}에서 {max}까지의 합 : {sum}")

#list = [i for i in range(min, max+1)]
#print(f"{min}에서 {max}까지의 합 : {sum(list)}")
```

```python
# min 부터 max 까지의 수를 출력하는 함수를 만들고, 1부터 10까지 출력하세요
# 함수명 : printNums
# 매개변수 : min, max
# 리턴값 : 없음
def printNums(min, max):
    # min이 max보다 크면 출력이 안됨 => 출력되게 하기 위해 min,max를 바꿈
    if min > max:
        min, max = max, min
    for i in range(min, max + 1):
        print(i)
printNums(1, 10)
# printNums(10, 1)
```
