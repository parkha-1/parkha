```python
# Phone 클래스를 추가
# 모델명(model), 가격(price)
# 기능 : 생성자 - 모델명과 가격을 초기화, 모델조회(display)

# IPhone 클래스를 추가(Phone 클래스를 상속)
# 페이스아이디적용여부(face_id) : boolean
# 기능 : 생성자 - 모델명과 가격, 페이스아이디 적용 여부를 초기화, 
#   모델조회(display) - 모델명, 가격, 페이스적용여부

# Galaxy 클래스를 추가(Phone 클래스를 상속)
# 펜 지원(s_pen)
# 기능 : 생성자 - 모델명, 가격, 펜지원여부를 초기화
#   모델조회(display) - 모델명, 가격, 펜지원 여부

# IPhone : iPhone15, 1250000, True
# IPhone : iPhoneSE, 650000, False
# Galaxy : S24 Ultra, 1690000, True
# Galaxy : Z Flip 5, 1390000, False


class Phone:
    def __init__ (self, model : str, price : int):
        self.model : str = model
        self.price : int = price

    def display(self):
        print(f"모델 : {self.model}\n가격 : {self.price:,}원")

class IPhone(Phone):
    def __init__(self, model : str, price : int, face_id : bool):
        super().__init__(model, price)
        self.face_id : bool = face_id

    def display(self):
        line_num = 20
        super().display()
        print(f"face id 적용여부: {"O" if self.face_id else "X"}")
        print("-" * line_num)

class Galaxy(Phone):
    def __init__(self, model : str, price : int, s_pen : bool):
        super().__init__(model, price)
        self.s_pen : bool = s_pen

    def display(self):
        line_num = 20
        super().display()
        print(f"펜 지원 여부: {"O" if self.s_pen else "X"}")
        print("-" * line_num)

c1 = IPhone("iPhone15", 1250000, True)
c1.display()

c1 = IPhone("iPhoneSE", 650000, False)
c1.display()

c2 = Galaxy("S24 Ultra", 1690000, True)
c2.display()

c2 = Galaxy("Z Flip 5", 1390000, False)
c2.display()

phones = [
    IPhone("iPhone15", 1250000, True),
    IPhone("iPhoneSE", 650000, False),
    Galaxy("S24 Ultra", 1690000, True),
    Galaxy("Z Flip 5", 1390000, False),
]

for phone in phones:
    phone.display()
```
